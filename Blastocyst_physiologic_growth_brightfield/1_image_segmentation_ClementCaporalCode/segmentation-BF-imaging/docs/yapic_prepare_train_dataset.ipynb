# Prepare the data set
The goal is to prepare a folder structure like this (where each img* is a 2D image)


```
yapic_training_dataset/
        └───input/
        │       img1.tiff
        │       img2.tiff
        │       img3.tiff
        │       ...
        │   
        └───target/
        │       img1_mask.tiff
        │       img2_mask.tiff
        │       img3_mask.tiff
        │       ...
```

You should create a original datafolder that has a similar layout (where each volume is a 3D volume (t, y, x))
this labels should be (t, y, x) with some time point completely annotated:

1. embryo
2. zona pelucida
3. background

```
3D_annotated_pair/
        └───raw/
        │       volume1.tiff
        │       volume2.tiff
        │       volume3.tiff
        │       ...
        │   
        └───label/
        │       volume1_mask.tiff
        │       volume2_mask.tiff
        │       volume3_mask.tiff
        │       ...
```

````python
import os
import numpy as np
import glob
from pathlib import Path
from tifffile import imread, imwrite
````

````python
folder_path_3D_annotated_pair = Path('../data/cell_discoverer_3D_annotated_pair')
folder_path_yapic_training_dataset = Path('../data/yapic_training_dataset')
````

````python
folder_path_3D_annotated_pair / "raw/*"
````

````python
path_imgs = glob.glob(str(folder_path_3D_annotated_pair / "raw/*"))
path_labels = glob.glob(str(folder_path_3D_annotated_pair / "label/*"))
````

````python
# create empty folder if it doesn't exist to save the data
if not os.path.isdir(folder_path_yapic_training_dataset):
    os.mkdir(folder_path_yapic_training_dataset)
    os.mkdir(folder_path_yapic_training_dataset/"input")
    os.mkdir(folder_path_yapic_training_dataset/"target")
````

````python
for k in range(len(path_imgs)):
    path_img = path_imgs[k]
    path_label = path_labels[k]
    print(path_img, path_label, "to", folder_path_yapic_training_dataset)
    im = imread(path_img)
    labels = imread(path_label)
    # populate the folders
    for i, t in enumerate(labels):
        if 1 in t: # if this slide has labels
            # save the pair
            imwrite(folder_path_yapic_training_dataset/"input"/(str(k)+"-"+str(i)+".tif"), [im[i].astype(np.uint8)], dtype=np.uint8)
            imwrite(folder_path_yapic_training_dataset/"target"/(str(k)+"-"+str(i)+".tif"), [t.astype(np.uint8)], dtype=np.uint8)
````
